### **05 - Product Analysis

## **Objetivo del analisis**

El objetivo de este notebook es comprender el desempeño de los productos dentro del ecommerce, identificando cuales generan mayor valor, cuales presentan mejores y peores resultados dentro del funnel de compra y que patrones de comportamiento pueden ser relevantes para  las decisiones comerciales.

A diferencia del analisis de clientes, donde el foco estuvo puesto en quien compra y como se comportan los usuarios, en este notebook el foco estara puesto en:
> ¿Que productos estan impulsando el negocio y como se comportan los clientes frente a ellos?
El analisis se realizara principalmente a nivel de product_id

## Objetivos especificos

### **1. Comprender el catalogo**

Determinar:
- Cuantos productos unicos existen
- Cuantos productos reciben visualizaciones
- cuantos productos llegan al carrito
- Cuantos productos registran compras
- Que proporcion del catalogo participa realmente en el funnel comercial

### **2. Identificare los productos que generan mayor valor**
Analizar el revenue generado por cada producto para identificar:
- Productos lideres en revenue
- Productos con menor contribucion
- Participacion porcentual de cada producto en el revenue total
- concentracion del revenue dentro del catalogo

El objetivo es determinar si el negocio depende de un grupo reducido de productos o si el revenue esta distribuido de manera mas equilibrada.

### **3. Analizar el comportamiento de los productos dentro del funnel**

Para cada producto se analizara su comportamiento a traves de:

#### **View to cart to purchase** como ya se habia hecho en previos pero esta vez a nivel de producto

Se calcularan metricas a nivel de producto como:
- view to cart rate
- cart to purchase rate
- view to purchase rate
Esto permitira distinguir entre productos que:
- Generan mucho trafico pero convierten poco
- Reciben poco trafico pero convierten bien
- presentan problemas principalmente en la etapa de carrito
- Tiene un desempeño fuerte a lo largo del funnel

Los rankings de conversion seran acompañados por criterios minimos de volumen para evitar que productos con muy pocas interacciones aparezcan artificialmente entre los mejores o peores resultados.

### **4. Analizar la recompra de productos**

se investigara que productos presentan mayores niveles de recompra por parte de los clientes.

La pregunta sera:
> ¿Que productos tienden a ser comprados nuevamente por los clientes que ya los adquirieron?
Esto permite identificar productos potencialmente relevantes para:
- Retencion
- Remarketing
- Reposicion
- Fidelizacion
- Estrategias de LTV
Los resultados seran interpretados considerando un volumen minimo de compradores para evitar conclusiones basadas en muestras demasiado pequeñas.

### **5. Analizar el comportamiento multi-producto**

Se estudiara cuantos productos son los comprados dentro de una misma sesion de compra y que productos aparecen asociados a sesiones donde los clientes adquieten multiples productos.

El objetivo es explorar posibles oportunidades de:
- Cross selling
- Aumento del numero de productos por compra
- Recomendaciones de productos
- Construccion de bundles
Este analisis sera exploratorio y no asumira automaticamente que la coexistencia de productos implica una relacion de compleentariedad.

### **6. Analizar la relacion entre el precio y desempeño**

De manera exploratiora, se analizara como se comportan los productos segun su nivel de precio.

Se buscara identificar patrones entre:
- Precio
- volumen de interaccion
- Conversion
- Revenue
- Comportamiento de compra
Este analisis permitira generar hipotesis comerciales, pero no se interpretaran como evidencia causal sobre el efecto del precio.

### **Principales preguntas de negocio**

A lo largo del notebook se buscara respnder:

- ¿Cuántos productos participan realmente en el funnel comercial?
- ¿Qué productos generan mayor revenue?
- ¿Qué tan concentrado está el revenue entre los productos?
- ¿Qué productos convierten mejor y cuáles presentan mayores dificultades?
- ¿Qué productos presentan mayores niveles de recompra?
- ¿Cuántos productos suelen comprar los clientes dentro de una misma sesión?
- ¿Qué productos aparecen con mayor frecuencia en sesiones multi-producto?
- ¿Existen diferencias relevantes en el comportamiento según el precio del producto?
- ¿Qué productos deberían recibir mayor atención comercial?
- ¿Qué oportunidades de Marketing, Ventas, CRO e inventario pueden derivarse del comportamiento observado?

### **Criterio de analisis**
Los productos no seran evaluados unicamente por una metrica.

Un producto con alto revenue no necesariamente es un producto con buena conversion, de la misma manera que un producto con una tasa de conversion elevada no necesariemnte representa una oportunidad comercial importante si tiene muy poco volumen.

Por ello, el analisis cobinara diferentes dimensiones:

**Volumen + conversion + revenue + recompra + comportamiento del cliente**

Esto permitira diferenciar entre:

- productos de alto volumen
- productos de alta eficiencia
- productos de alto valor economico
- productos con potencial de recompra
- productos con posibles oportunidades de mejora

### **Limitaciones relacionadas con las categorias**

La columna category_code presenta un porcentaje muy elevado de valores faltantes, aproximadamente 98.35%, con un patrón estructural asociado a determinados productos.

Por esta razón, no se realizará un análisis de desempeño a nivel de categoría en este notebook.

El análisis se concentrará en product_id, donde existe información suficiente para realizar comparaciones y construir rankings de desempeño.

Al finalizar el notrebook se espera obtener una vision integral del desempeño del catalogo y transformar los resultados en:

Evidencia → Insight → Implicación de negocio → Oportunidad → Acción propuesta

el objetivo final no es simplemente identificar cuales son los productos "buenos" o "malos", si no en comprender por que presentan determinado comportamiento y que decisiones empresariales podrian tomarse a partir de esa informacion.
