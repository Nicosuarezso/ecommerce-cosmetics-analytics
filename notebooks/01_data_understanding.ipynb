# 1. Conexion a la base de datos
El objetivo de esta seccion es entender la estructura de la base de datos antes de cargarla o empezar directo con el analisis

```python
# Importacion de librerias

import sqlite3
import pandas as pd
import numpy as np

DB_PATH = "../data/raw/ecommerce.db"

conn = sqlite3.connect(DB_PATH)
```

```python
query = """
SELECT name
FROM sqlite_master
WHERE type='table';"""

tables = pd.read_sql_query(query, conn)

tables
```

Inspeccionar la estructura de la tabla 2019-Oct para entender el schema antes de evaluar las otras tablas

```python
query = """
PRAGMA table_info('2019-Oct');"""

oct_schema = pd.read_sql_query(query, conn)

oct_schema
```

```python
sample_query = """
SELECT *
FROM "2019-Oct"
LIMIT 10;
"""

sample = pd.read_sql_query(sample_query, conn)

sample
```

```python
event_query = """
SELECT
    event_type,
    COUNT(*) AS event_count
FROM "2019-Oct"
GROUP BY event_type
ORDER BY event_count DESC;
"""

event_types = pd.read_sql_query(event_query, conn)

event_types
```

Comprobacion de todas las tablas

```python
tables = [
    "2019-Oct",
    "2019-Nov",
    "2019-Dec",
    "2020-Jan",
    "2020-Feb"
]

schemas = {}

for table in tables:
    query = f'PRAGMA table_info("{table}");'
    schemas[table] = pd.read_sql_query(query, conn)

schemas
```

```python
for table, schema in schemas.items():
    print(f"\n{table}")
    print(schema["name"].tolist())
```

```python
# Cuantos eventos contiene cada tabla 

row_counts = {}

for table in tables:
    query = f'SELECT COUNT(*) AS row_count FROM "{table}";'
    result = pd.read_sql_query(query, conn)
    row_counts[table] = result.loc[0, "row_count"]

row_counts
```

```python
table_counts = pd.DataFrame(
    list(row_counts.items()),
    columns=["table", "row_count"]
)

total_rows = table_counts["row_count"].sum()

total_rows
```

```python
table_counts
```

```python
date_ranges = []

for table in tables:
    query = f"""
    SELECT
        MIN(event_time) AS min_event_time,
        MAX(event_time) AS max_event_time
    FROM "{table}";
    """
    
    result = pd.read_sql_query(query, conn)
    
    date_ranges.append({
        "table": table,
        "min_event_time": result.loc[0, "min_event_time"],
        "max_event_time": result.loc[0, "max_event_time"]
    })

date_ranges = pd.DataFrame(date_ranges)

date_ranges
```
